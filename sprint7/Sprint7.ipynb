## Algoritmo Lenguaje Natural (Pensamiento Modular)
El problema esta dividido en 6 funciones independientes y una que la coordina a todas.

**validar_datos:** Comprueba que los años de experiencia sean mayores o iguales a 0, y que las puntuaciones de Catalán, Excel y Titulación estén entre 0 y 3. Devuelve True si todo es correcto o False si hay algún error.

**calcular_puntuacion:** Multiplica los años de experiencia por 1.5 y le suma los puntos de Catalán, Excel y Titulación para obtener la nota final.

**clasificar_candidato:** Evalúa la puntuación recibida y asigna la categoría correspondiente:

Mayor o igual a 12 → Prioridad Alta

Entre 8 y 11.99 → Candidato adecuado

Menor a 8 → No prioritario

**mostrar_individual:** Muestra un mensaje en pantalla tras procesar a cada candidato, indicando su nombre, puntuación y clasificación.

**mostrar_informe_dataframe:** Toma la lista de candidatos procesados, la convierte en un DataFrame de pandas, ordena los registros según la prioridad (Prioridad Alta > Candidato adecuado > No prioritario) y, en caso de empate, por la puntuación de mayor a menor. Muestra la tabla final en pantalla.

**principal_seleccion:** Función coordinadora. Recorre los datos de prueba en bloques de 6 elementos, llama a la validación, procesa únicamente a los candidatos válidos y se detiene exactamente al alcanzar 10 inscritos válidos. Por último, llama al informe en DataFrame.


```python
# SPRINT 7 - NIVEL 1 - EJERCICIO 1: SELECCIÓN DE PERSONAL (CON DATAFRAME)

import pandas as pd

def validar_datos(experiencia, catalan, excel, titulacion):
    """Comprueba que todos los datos numéricos estén dentro de los rangos permitidos."""
    if experiencia < 0:
        return False
    if catalan not in [0, 1, 2, 3]:
        return False
    if excel not in [0, 1, 2, 3]:
        return False
    if titulacion not in [0, 1, 2, 3]:
        return False
    return True


def calcular_puntuacion(experiencia, catalan, excel, titulacion):
    """Calcula el puntaje total aplicando el peso de 1.5 a la experiencia."""
    puntos_experiencia = experiencia * 1.5
    puntuacion_total = puntos_experiencia + catalan + excel + titulacion
    return puntuacion_total


def clasificar_candidato(puntuacion):
    """Devuelve la etiqueta de prioridad según la puntuación obtenida."""
    if puntuacion >= 12:
        return "Prioridad Alta"
    elif puntuacion >= 8:
        return "Candidato adecuado"
    else:
        return "No prioritario"


def mostrar_individual(nombre, puntuacion, categoria):
    """Muestra el resultado individual en pantalla tras registrar a cada candidato."""
    print(f"✅ Candidato/a registrado/a: {nombre} | Puntuación: {puntuacion:.2f} | Categoría: {categoria}")


def mostrar_informe_dataframe(lista_candidatos):
    """
    Convierte la lista de candidatos en un DataFrame de pandas,
    ordena según los criterios requeridos por RRHH e imprime la tabla final.
    """
    # 1. Crear el DataFrame
    df_candidatos = pd.DataFrame(lista_candidatos)

    # 2. Crear una columna auxiliar para poder ordenar numéricamente por la categoría
    mapa_prioridad = {"Prioridad Alta": 1, "Candidato adecuado": 2, "No prioritario": 3}
    df_candidatos["orden_prioridad"] = df_candidatos["categoria"].map(mapa_prioridad)

    # 3. Ordenar: primero por prioridad (1 -> 2 -> 3) y en caso de empate por puntuación (descendente)
    df_ordenado = df_candidatos.sort_values(
        by=["orden_prioridad", "puntuacion"],
        ascending=[True, False]
    ).reset_index(drop=True)

    # 4. Eliminar la columna auxiliar para dejar el informe limpio
    df_informe = df_ordenado.drop(columns=["orden_prioridad"])

    # Ajustamos el índice para que empiece en 1 (Ranking)
    df_informe.index = df_informe.index + 1
    df_informe.index.name = "Posición"
    print("📋 INFORME FINAL DE SELECCIÓN DE PERSONAL (TOP 10)")
    print(df_informe.to_string())
    return df_informe


def principal_seleccion(datos_entrada):
    """Función principal que orquesta el proceso completo de inscripción."""
    candidatos_validos = []
    indice = 0
    total_datos = len(datos_entrada)

    # Procesamos hasta completar 10 candidatos válidos o agotar los datos recibidos
    while len(candidatos_validos) < 10 and indice < total_datos:
        nombre = datos_entrada[indice]
        telefono = datos_entrada[indice + 1]
        experiencia = datos_entrada[indice + 2]
        catalan = datos_entrada[indice + 3]
        excel = datos_entrada[indice + 4]
        titulacion = datos_entrada[indice + 5]

        indice += 6  # Avanzamos 6 posiciones para leer el siguiente candidato

        # Validación de datos
        if not validar_datos(experiencia, catalan, excel, titulacion):
            print(f"⚠️ REGISTRO RECHAZADO: Datos no válidos para '{nombre}'. Se omite este registro.")
            continue

        # Cálculo y clasificación
        puntuacion = calcular_puntuacion(experiencia, catalan, excel, titulacion)
        categoria = clasificar_candidato(puntuacion)

        # Guardar candidato válido en el listado
        candidatos_validos.append({
            "nombre": nombre,
            "telefono": telefono,
            "puntuacion": puntuacion,
            "categoria": categoria
        })

        # Mostrar resultado individual en pantalla
        mostrar_individual(nombre, puntuacion, categoria)

    # Generar el informe final presentado como DataFrame
    df_final = mostrar_informe_dataframe(candidatos_validos)
    return df_final
```

Descomposición de funciones y responsabilidades:
obtener_segmento_y_factor(codigo):

Objetivo: Traducir el código de comportamiento numérico (0 al 4) a información de negocio sin usar estructuras if/elif.

Pasos: Consulta un diccionario donde la clave es el número del código y devuelve una tupla con tres elementos: la descripción del segmento, el mensaje publicitario recomendado para la campaña y el factor de código numérico.

Objetivo: Obtener el multiplicador correspondiente al rango de edad del cliente.

Pasos: Utiliza un diccionario directo donde las claves son las cadenas de texto del grupo de edad  y devuelve el factor flotante correspondiente.

```python
# SPRINT 7 - NIVEL 1 - EJERCICIO 2: MARKETING DIGITAL

import pandas as pd

def obtener_segmento_y_factor(codigo):
    """
    Busca en un diccionario la información del comportamiento sin utilizar condicionales if/elif.
    Devuelve una tupla: (segmento, mensaje_publicitario, factor_codigo)
    """
    tabla_comportamiento = {
        0: ("Sin intención de compra", "Recomendar productos más vendidos de la tienda", 0.10),
        1: ("Intención inicial abandonada", "Ofrecer un 5% de descuento en los productos guardados", 0.25),
        2: ("Abandono en checkout", "Recordar métodos de pago disponibles y envío gratuito", 0.50),
        3: ("Abandono reciente (<24h)", "Recordatorio amigable de carrito pendiente", 0.75),
        4: ("Abandono crítico (>24h)", "Ofrecer un cupón de urgencia del 10% de descuento", 0.90)
    }
    return tabla_comportamiento.get(codigo, ("Desconocido", "Sin mensaje", 0.0))


def obtener_factor_edad(grupo_edad):
    """Busca el factor multiplicador de edad en un diccionario directo."""
    tabla_edad = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }
    return tabla_edad.get(grupo_edad, 1.0)


def calcular_riesgo(factor_codigo, factor_edad):
    """Calcula la puntuación de riesgo final multiplicando ambos factores."""
    return round(factor_codigo * factor_edad, 4)


def procesar_clientes(lista_clientes):
    """Recorre la lista de clientes y calcula todas las métricas de negocio requeridas."""
    clientes_procesados = []

    for cliente in lista_clientes:
        segmento, mensaje, factor_codigo = obtener_segmento_y_factor(cliente["codigo_comportamiento"])
        factor_edad = obtener_factor_edad(cliente["grupo_edad"])
        riesgo_final = calcular_riesgo(factor_codigo, factor_edad)

        # Creamos una copia del registro y agregamos las nuevas variables procesadas
        cliente_enriquecido = cliente.copy()
        cliente_enriquecido.update({
            "segmento": segmento,
            "mensaje": mensaje,
            "factor_codigo": factor_codigo,
            "factor_edad": factor_edad,
            "riesgo_final": riesgo_final
        })
        clientes_procesados.append(cliente_enriquecido)

    return clientes_procesados


def mostrar_dataframe(clientes_procesados):
    """Convierte la lista procesada en un DataFrame de pandas para una visualización tabular."""
    df = pd.DataFrame(clientes_procesados)

    # Seleccionamos las columnas para la visualización en pantalla
    columnas_vista = [
        "id_cliente", "nombre", "edad", "grupo_edad",
        "codigo_comportamiento", "factor_codigo", "factor_edad",
        "riesgo_final", "segmento"
    ]

    print("\n--- DATASET PROCESADO DE CLIENTES ---")
    print(df[columnas_vista].to_string(index=False))
    return df


def mostrar_respuestas(df_clientes):
    """Responde con argumentos analíticos a las 4 preguntas planteadas por Marketing."""
    print("\n" + "=" * 75)
    print("INFORME ANALÍTICO PARA EL DEPARTAMENTO DE MARKETING")
    print("=" * 75)

    # Pregunta 1: Clientes a contactar primero en retargeting
    df_ordenado_riesgo = df_clientes.sort_values(by="riesgo_final", ascending=False)
    print("\n Pregunta 1 Clientes que deben contactarse PRIMERO (Mayor riesgo de abandono):")
    for idx, fila in df_ordenado_riesgo.head(3).iterrows():
        print(f"   • {fila['nombre']} (Riesgo: {fila['riesgo_final']:.2f} | Grupo Edad: {fila['grupo_edad']} | Código: {fila['codigo_comportamiento']})")

    # Pregunta 2: Diferencias entre mismo comportamiento y distintas edades
    print("\n Pregunta 2 Impacto del factor EDAD para un mismo comportamiento de compra:")
    print("   • El grupo de edad 18-25 años aplica un multiplicador de 1.20, mientras que el grupo 60+ aplica 0.70.")
    print("   • Ejemplo práctico: Laura Serra (23 años, código 4) alcanza un riesgo de 1.08, mientras que Mónica Pérez (47 años, código 4) alcanza 0.77. El público joven presenta una mayor volatilidad de abandono.")

    # Pregunta 3: Clientes con poco interés donde no conviene invertir presupuesto
    clientes_bajo_interes = df_clientes[df_clientes["riesgo_final"] < 0.20]
    print("\n Pregunta 3 Clientes NO PRIORITARIOS (Bajo interés / Riesgo < 0.20):")
    for idx, fila in clientes_bajo_interes.iterrows():
        print(f"   • {fila['nombre']} (Riesgo: {fila['riesgo_final']:.2f} | Código: {fila['codigo_comportamiento']}) - Motivo: Solo navega por la web sin agregar productos.")

    # Pregunta 4: Cliente más rentable de recuperar
    clientes_checkout = df_clientes[df_clientes["codigo_comportamiento"] == 2].sort_values(by="riesgo_final", ascending=False)
    mas_rentable = clientes_checkout.iloc[0]
    print("\n Pregunta 4 Cliente MÁS RENTABLE de recuperar:")
    print(f"   • {mas_rentable['nombre']} (ID Cliente: {mas_rentable['id_cliente']})")
    print(f"     Explicación: Llegó hasta el paso final de pago en checkout (Código 2), demostrando intención de compra inmediata. Al pertenecer al rango de {mas_rentable['grupo_edad']} años, su riesgo de pérdida es {mas_rentable['riesgo_final']:.2f}. Un recordatorio rápido garantiza la conversión.")
    print("=" * 75)


def principal_marketing(lista_clientes):
    """Función principal orquestadora del análisis de Marketing."""
    datos_procesados = procesar_clientes(lista_clientes)
    df_resultado = mostrar_dataframe(datos_procesados)
    mostrar_respuestas(df_resultado)
```

#Respuesta Nivel 2 Ejercicio 1
Ejercicio 1: Gestión de riesgo y priorización de pacientes (Centro Médico)
1. Pensamiento Modular y Algoritmo en Lenguaje Natural
En este ejercicio transformamos registros de texto desordenados en un DataFrame organizado y priorizado para el equipo médico. Diseñamos un flujo modular con 5 funciones donde cada una resuelve una única tarea:

limpiar_y_normalizar_sintomas(cadena_sintomas):

Objetivo: Estandarizar la cadena de síntomas enviada por el paciente para que coincida con las claves del diccionario de riesgos.

Pasos: Convierte el texto a minúsculas, divide la cadena por comas (,), elimina espacios sobrantes en las orillas de cada síntoma y reemplaza los espacios internos por guiones bajos (_).

* calcular_riesgo_paciente(lista_sintomas, diccionario_riesgos):

Objetivo: Obtener la puntuación de gravedad total acumulada.

Pasos: Recorre la lista de síntomas limpios, busca el valor numérico correspondiente en el diccionario niveles_riesgo y suma los puntos.

* procesar_pacientes_a_dataframe(lista_pacientes_raw, diccionario_riesgos):

Objetivo: Parsear los textos originales y estructurarlos en una tabla.

Pasos: Separa el string del paciente usando el delimitador | para extraer el nombre, la edad y la cadena de síntomas. Invoca a las dos subfunciones anteriores y construye un DataFrame de pandas con las columnas: nombre, edad, sintomas_normalizados y riesgo_total.

* ordenar_pacientes_por_prioridad(df_pacientes):

Objetivo: Determinar el orden de atención médica en urgencias.

Pasos: Ordena el DataFrame de mayor a menor según la columna riesgo_total y reajusta el índice para asignar las posiciones de prioridad (1, 2, 3...).

* principal_pacientes(lista_pacientes_raw, diccionario_riesgos):

Objetivo: Orquestar el flujo de triaje e imprimir el informe médico en pantalla.

```python
# SPRINT 7 - NIVEL 2 - EJERCICIO 1: TRIAGE Y PACIENTES

import pandas as pd

def limpiar_y_normalizar_sintomas(cadena_sintomas):
    """
    Convierte la cadena de síntomas a minúsculas, quita espacios extra
    y sustituye espacios por guiones bajos para emparejar con el diccionario de riesgos.
    """
    lista_cruda = cadena_sintomas.lower().split(",")
    sintomas_limpios = []

    for sintoma in lista_cruda:
        # Quitamos espacios extremos y reemplazamos espacios por '_'
        sintoma_formateado = sintoma.strip().replace(" ", "_")
        sintomas_limpios.append(sintoma_formateado)

    return sintomas_limpios


def calcular_riesgo_paciente(lista_sintomas, diccionario_riesgos):
    """Suma la puntuación de riesgo correspondiente a cada síntoma del paciente."""
    riesgo_total = 0
    for sintoma in lista_sintomas:
        # Buscamos el síntoma; si no existe, asignamos 0 puntos por defecto
        riesgo_total += diccionario_riesgos.get(sintoma, 0)
    return riesgo_total


def procesar_pacientes_a_dataframe(lista_pacientes_raw, diccionario_riesgos):
    """Parsea el listado de cadenas desestructuradas y crea un DataFrame organizado."""
    registros_pacientes = []

    for registro in lista_pacientes_raw:
        # Separamos los 3 bloques divididos por '|'
        partes = registro.split("|")
        nombre = partes[0].strip()
        edad = int(partes[1].strip())
        cadena_sintomas = partes[2]

        # Limpieza de síntomas y cálculo de riesgo
        sintomas_normalizados = limpiar_y_normalizar_sintomas(cadena_sintomas)
        riesgo_total = calcular_riesgo_paciente(sintomas_normalizados, diccionario_riesgos)

        # Guardamos la ficha del paciente
        registros_pacientes.append({
            "nombre": nombre,
            "edad": edad,
            "sintomas_normalizados": ", ".join(sintomas_normalizados),
            "riesgo_total": riesgo_total
        })

    # Creamos y devolvemos el DataFrame
    df_pacientes = pd.DataFrame(registros_pacientes)
    return df_pacientes


def ordenar_pacientes_por_prioridad(df_pacientes):
    """Ordena a los pacientes descendentemente por riesgo total y ajusta el ranking."""
    df_ordenado = df_pacientes.sort_values(
        by="riesgo_total",
        ascending=False
    ).reset_index(drop=True)

    # Ajustamos el índice para mostrar las posiciones de prioridad (1, 2, 3...)
    df_ordenado.index = df_ordenado.index + 1
    df_ordenado.index.name = "Prioridad"
    return df_ordenado


def principal_pacientes(lista_pacientes_raw, diccionario_riesgos):
    """Función principal orquestadora del centro médico."""

    print(" INFORME DE TRIAJE Y PRIORIZACIÓN DE PACIENTES")


    # 1. Procesar datos a DataFrame
    df_procesado = procesar_pacientes_a_dataframe(lista_pacientes_raw, diccionario_riesgos)

    # 2. Ordenar por prioridad médica
    df_priorizado = ordenar_pacientes_por_prioridad(df_procesado)

    # 3. Mostrar tabla
    print(df_priorizado.to_string())


    # 4. Alerta médica inmediata
    primer_paciente = df_priorizado.iloc[0]
    print(f" PRIMER PACIENTE A ATENDER: {primer_paciente['nombre']} (Riesgo Total: {primer_paciente['riesgo_total']} puntos)")
    return df_priorizado
```

```python
# Datos de entrada traducidos al castellano
pacientes_prueba = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

# Ejecutamos la función principal
df_resultado_pacientes = principal_pacientes(pacientes_prueba, niveles_riesgo)
```

1. Pensamiento Modular y Algoritmo en Lenguaje Natural

Para resolver este problema sin utilizar estructuras condicionales if/elif y mantener un diseño modular limpio, estructuramos el sistema en 5 funciones con responsabilidades bien delimitadas:

* obtener_datos_categoria(codigo, diccionario_niveles):

Objetivo: Mapear la información de riesgo y costo según el código de comportamiento del cliente sin condicionales.

Pasos: Consulta directamente el diccionario niveles_seguro usando la clave numérica codigo y devuelve una tupla con la descripción de la categoría, el precio base y la probabilidad de fraude.

* calcular_precio_final(precio_base, edad):

Objetivo: Aplicar la fórmula financiera de la póliza.

Pasos: Calcula $\text{precio\_final} = \text{precio\_base} \times \left(1 + \frac{\text{edad}}{100}\right)$ y redondea el resultado a 2 decimales.

* crear_dataframe_seguros(lista_clientes, diccionario_niveles):

Objetivo: Transformar el listado original de clientes en un DataFrame enriquecido.

Pasos: Recorre la lista de clientes, invoca a las funciones obtener_datos_categoria y calcular_precio_final, y construye un DataFrame de pandas con todas las variables consolidadas.

* analizar_y_ordenar_seguros(df_seguros):

Objetivo: Generar las vistas ordenadas y responder a las preguntas de análisis de riesgo de la aseguradora.

Pasos: Ordena el DataFrame por probabilidad_fraude en orden descendente.
Ordena el DataFrame por precio_final en orden descendente.Identifica y presenta en pantalla al cliente con mayor riesgo de fraude, al cliente con la póliza más costosa y al candidato prioritario para revisión manual.

* principal_seguros(lista_clientes, diccionario_niveles):

Objetivo: Orquestar el flujo completo de análisis actuarial.

```python
# SPRINT 7 - NIVEL 2 - EJERCICIO 2: COMPAÑÍA ASEGURADORA


import pandas as pd

def obtener_datos_categoria(codigo, diccionario_niveles):
    """
    Busca las métricas asociadas al código en el diccionario de niveles sin utilizar if/elif.
    Devuelve: (categoria, precio_base, probabilidad_fraude)
    """
    datos = diccionario_niveles.get(
        codigo,
        {"categoria": "Desconocido", "precio_base": 0, "fraude": 0.0}
    )
    return datos["categoria"], datos["precio_base"], datos["fraude"]


def calcular_precio_final(precio_base, edad):
    """Aplica la fórmula: precio_final = precio_base * (1 + edad / 100)."""
    precio_final = precio_base * (1 + (edad / 100))
    return round(precio_final, 2)


def crear_dataframe_seguros(lista_clientes, diccionario_niveles):
    """Procesa la lista de clientes y devuelve un DataFrame con la información calculada."""
    registros_clientes = []

    for cliente in lista_clientes:
        categoria, precio_base, prob_fraude = obtener_datos_categoria(
            cliente["codigo"],
            diccionario_niveles
        )
        precio_final = calcular_precio_final(precio_base, cliente["edad"])

        registros_clientes.append({
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "codigo": cliente["codigo"],
            "categoria": categoria,
            "precio_base": precio_base,
            "precio_final": precio_final,
            "probabilidad_fraude": prob_fraude
        })

    # Construcción del DataFrame de pandas
    df_seguros = pd.DataFrame(registros_clientes)
    return df_seguros


def analizar_y_ordenar_seguros(df_seguros):
    """Genera los informes ordenados y extrae las conclusiones de negocio."""
    print("\n--- DATAFRAME COMPLETO DE CLIENTES DE SEGUROS ---")
    columnas_mostrar = ["nombre", "edad", "categoria", "precio_base", "precio_final", "probabilidad_fraude"]
    print(df_seguros[columnas_mostrar].to_string(index=False))

    # 1. Ordenación por probabilidad de fraude (descendente)
    df_orden_fraude = df_seguros.sort_values(by="probabilidad_fraude", ascending=False)

    # 2. Ordenación por precio final (descendente)
    df_orden_precio = df_seguros.sort_values(by="precio_final", ascending=False)


    print(" ANÁLISIS DE RIESGO Y REVISIÓN MANUAL")


    cliente_max_fraude = df_orden_fraude.iloc[0]
    cliente_max_precio = df_orden_precio.iloc[0]

    print(f" Cliente con MÁS PROBABILIDAD DE FRAUDE:")
    print(f"   • {cliente_max_fraude['nombre']} (Probabilidad: {cliente_max_fraude['probabilidad_fraude']*100:.0f}% | Categoría: {cliente_max_fraude['categoria']})")

    print(f" Cliente con el PRECIO FINAL MÁS ALTO:")
    print(f"   • {cliente_max_precio['nombre']} (Precio Final: {cliente_max_precio['precio_final']} € | Edad: {cliente_max_precio['edad']} años)")

    print(f" Cliente PRIORITARIO para revisión manual:")
    print(f"   • {cliente_max_fraude['nombre']}: Presenta un nivel de 'Riesgo crítico' con un {cliente_max_fraude['probabilidad_fraude']*100:.0f}% de probabilidad de fraude, por lo que se debe auditar su solicitud antes de emitir la póliza.")



def principal_seguros(lista_clientes, diccionario_niveles):
    """Función principal orquestadora del análisis de seguros."""
    df_seguros = crear_dataframe_seguros(lista_clientes, diccionario_niveles)
    analizar_y_ordenar_seguros(df_seguros)
    return df_seguros
```

1. Pensamiento Modular y Algoritmo en Lenguaje Natural
Como analistas en el departamento de Ciberseguridad, abordamos el problema de las contraseñas débiles en dos etapas bien diferenciadas:

Auditoría de vulnerabilidades: Analizamos la contraseña actual del empleado para identificar exactamente qué políticas de seguridad está incumpliendo (longitud, falta de símbolos, etc.).

Generación de reemplazo seguro: Si la clave es insegura, la descartamos y generamos una totalmente nueva mediante una función paramétrica modular que garantiza la presencia de los tipos de caracteres seleccionados y aleatoriza sus posiciones finales con la librería numpy.

Descomposición de funciones y responsabilidades:

* obtener_bancos_caracteres():


Objetivo: Centralizar las cadenas con los caracteres disponibles (mayúsculas, minúsculas, números y símbolos) en un diccionario estructurado.

* analizar_contrasena_existente(contrasena_actual):

Objetivo: Evaluar la clave débil del empleado e identificar sus fallos.


Pasos: Revisa si la longitud es menor a 12 caracteres y si le faltan mayúsculas, minúsculas, números o caracteres especiales. Devuelve True si es segura o False si presenta vulnerabilidades.


* garantizar_caracteres_minimos(majuscules, minuscules, numeros, signes, bancos):

Objetivo: Asegurar que la nueva contraseña contenga obligatoriamente al menos un carácter de cada categoría activada.

Pasos: Utiliza np.random.choice() sobre cada conjunto activado para extraer un elemento inicial y construye el grupo total de caracteres permitidos.

* crear_contrasena(longitud=12, majuscules=True, minuscules=True, numeros=True, signes=False):

Objetivo: Función principal del generador con las opciones por defecto requeridas.

Pasos:

Llama a garantizar_caracteres_minimos() para obtener la base de la clave.

Rellena las posiciones faltantes hasta alcanzar la longitud deseada seleccionando caracteres al azar del grupo permitido mediante np.random.choice().

Mezcla el orden total de los caracteres con np.random.shuffle() para evitar patrones predecibles.

Une la lista y devuelve la contraseña en texto.

* flujo_trabajo_ciberseguridad(contrasena_empleado):

Objetivo: Función orquestadora que une la auditoría de la clave vieja con la asignación de la nueva contraseña segura.

```python
# SPRINT 7 - NIVEL 3 - EJERCICIO 1: CIBERSEGURIDAD


import numpy as np
import string

def obtener_bancos_caracteres():
    """Devuelve un diccionario con las cadenas de caracteres disponibles por categoría."""
    return {
        "mayusculas": string.ascii_uppercase,
        "minusculas": string.ascii_lowercase,
        "numeros": string.digits,
        "signes": "!@#$%^&*()_+-=[]{}|;:,.<>?"
    }


def analizar_contrasena_existente(contrasena_actual):
    """
    Evalúa la contraseña actual del empleado e identifica sus vulnerabilidades.
    Devuelve True si cumple todas las reglas o False si es insegura.
    """
    bancos = obtener_bancos_caracteres()
    vulnerabilidades = []

    if len(contrasena_actual) < 12:
        vulnerabilidades.append(f"Longitud insuficiente ({len(contrasena_actual)} caracteres; mínimo requerido: 12)")
    if not any(caracter in bancos["mayusculas"] for caracter in contrasena_actual):
        vulnerabilidades.append("Carece de letras mayúsculas")
    if not any(caracter in bancos["minusculas"] for caracter in contrasena_actual):
        vulnerabilidades.append("Carece de letras minúsculas")
    if not any(caracter in bancos["numeros"] for caracter in contrasena_actual):
        vulnerabilidades.append("Carece de números")
    if not any(caracter in bancos["signes"] for caracter in contrasena_actual):
        vulnerabilidades.append("Carece de caracteres especiales / símbolos")

    print(f"AUDITORÍA DE SEGURIDAD DE LA CLAVE ACTUAL: '{contrasena_actual}'")
    if vulnerabilidades:
        print("ESTADO DE LA CLAVE: INSEGURA / VULNERABLE")
        print("   Vulnerabilidades detectadas:")
        for fallo in vulnerabilidades:
            print(f"   • {fallo}")
        return False
    else:
        print("ESTADO DE LA CLAVE: SEGURA (Cumple con las políticas corporativas)")
        return True


def garantizar_caracteres_minimos(mayusculas, minusculas, numeros, signes, bancos):
    """Garantiza al menos un carácter inicial al azar por cada tipo seleccionado."""
    obligatorios = []
    pool_permitido = ""

    if mayusculas:
        obligatorios.append(np.random.choice(list(bancos["mayusculas"])))
        pool_permitido += bancos["mayusculas"]

    if minusculas:
        obligatorios.append(np.random.choice(list(bancos["minusculas"])))
        pool_permitido += bancos["minusculas"]

    if numeros:
        obligatorios.append(np.random.choice(list(bancos["numeros"])))
        pool_permitido += bancos["numeros"]

    if signes:
        obligatorios.append(np.random.choice(list(bancos["signes"])))
        pool_permitido += bancos["signes"]

    return obligatorios, pool_permitido


def crear_contrasena(longitud=12, mayusculas=True, minusculas=True, numeros=True, signes=False):
    """
    Función principal para generar contraseñas seguras utilizando numpy.

    Valores por defecto de la política corporativa:
    - longitud = 12
    - mayusculas = True
    - minusculas = True
    - numeros = True
    - signes = False
    """
    bancos = obtener_bancos_caracteres()

    # 1. Obtener caracteres mínimos requeridos y pool total permitido
    obligatorios, pool_permitido = garantizar_caracteres_minimos(
        mayusculas, minusculas, numeros, signes, bancos
    )

    if not pool_permitido:
        raise ValueError("Debes seleccionar al menos un tipo de carácter.")

    if longitud < len(obligatorios):
        longitud = len(obligatorios)

    # 2. Rellenar los caracteres restantes al azar
    faltantes_count = longitud - len(obligatorios)
    restantes = list(np.random.choice(list(pool_permitido), size=faltantes_count))

    # 3. Combinar y aleatorizar el orden total con numpy
    lista_caracteres = obligatorios + restantes
    np.random.shuffle(lista_caracteres)

    # 4. Convertir a string continuo
    return "".join(lista_caracteres)


def flujo_trabajo_ciberseguridad(contrasena_empleado):
    """Orquesta el proceso de auditoría y reemplazo automatizado."""

    print("DEPARTAMENTO DE CIBERSEGURIDAD - AUDITORÍA Y REEMPLAZO")

    # 1. Analizar la clave vulnerable del empleado
    es_segura = analizar_contrasena_existente(contrasena_empleado)

    # 2. Si la clave es insegura, generar inmediatamente la nueva contraseña
    if not es_segura:
        print("Generando nueva contraseña de reemplazo según estándar corporativo...")
        nueva_clave = crear_contrasena(12, mayusculas=True, minusculas=True, numeros=True, signes=True)
        print(f"NUEVA CONTRASEÑA SEGURA ASIGNADA: {nueva_clave}")
```

#Ejercicio 2:
Procesamiento automático de datos deportivos (Liga Catalana)
1. Pensamiento Modular y Algoritmo en Lenguaje NaturalAjustamos la salida del DataFrame segun imagen del ejercicio:   

Índice del DataFrame: Nombre del equipo (equipo).  

Columnas de la tabla: goles_a_favor, goles_en_contra, ganados, empatados, perdidos, puntos.   

Orden de las columnas: Se coloca primero la estadística de goles y partidos antes de los puntos finales.   
Ordenación de filas: Ordenado de forma descendente por la columna puntos.   

Descomposición de funciones y responsabilidades:

* leer_archivo_partidos(ruta_archivo): Lee las líneas del archivo .txt eliminando espacios sobrantes.  

* parsear_linea_partido(linea): Extrae el equipo local, sus goles, el equipo visitante y sus goles.   

* actualizar_estadisticas_equipo(tabla, equipo, goles_favor, goles_contra): Incrementa los contadores acumulados de partidos y puntos.   

* generar_clasificacion_dataframe(lineas): Procesa las líneas, construye el DataFrame, asigna equipo como índice y reordena las columnas tal como exige la plataforma.   

* mostrar_informe_deportivo(df_clasificacion): Muestra la tabla resultante y responde al formato de texto esperado (Equipo más goleador:, Equipo más goleado:, etc.).   

```python
# SPRINT 7 - NIVEL 3 - EJERCICIO 2: LIGA CATALANA


import pandas as pd

def leer_archivo_partidos(ruta_archivo="historic partits.txt"):
    """Lee el archivo de texto y devuelve una lista con las líneas limpias."""
    try:
        with open(ruta_archivo, "r", encoding="utf-8") as archivo:
            return [linea.strip() for linea in archivo if linea.strip()]
    except FileNotFoundError:
        print(f"❌ Error: No se encontró el archivo '{ruta_archivo}'.")
        return []


def parsear_linea_partido(linea):
    """
    Parsea una línea con formato 'EquipoLocal\tGolesL-GolesV\tEquipoVisitante'.
    Devuelve: (equipo_local, goles_local, equipo_visitante, goles_visitante)
    """
    partes = linea.split("\t")
    equipo_local = partes[0].strip()

    # Marcador central (ejemplo: '4-5')
    marcador = partes[1].strip().split("-")
    goles_local = int(marcador[0])
    goles_visitante = int(marcador[1])

    equipo_visitante = partes[2].strip()

    return equipo_local, goles_local, equipo_visitante, goles_visitante


def inicializar_estadisticas():
    """Devuelve la estructura de contadores iniciales para un club."""
    return {
        "goles_a_favor": 0,
        "goles_en_contra": 0,
        "ganados": 0,
        "empatados": 0,
        "perdidos": 0,
        "puntos": 0
    }


def actualizar_estadisticas_equipo(tabla, equipo, goles_favor, goles_contra):
    """Acumula las estadísticas de un equipo tras disputar un partido."""
    if equipo not in tabla:
        tabla[equipo] = inicializar_estadisticas()

    stats = tabla[equipo]
    stats["goles_a_favor"] += goles_favor
    stats["goles_en_contra"] += goles_contra

    # Asignación de puntos y contadores de partido
    if goles_favor > goles_contra:
        stats["puntos"] += 3
        stats["ganados"] += 1
    elif goles_favor == goles_contra:
        stats["puntos"] += 1
        stats["empatados"] += 1
    else:
        stats["perdidos"] += 1


def generar_clasificacion_dataframe(lineas):
    """Procesa las líneas y devuelve el DataFrame con el formato de la plataforma."""
    tabla_equipos = {}

    for linea in lineas:
        eq_loc, g_loc, eq_vis, g_vis = parsear_linea_partido(linea)
        actualizar_estadisticas_equipo(tabla_equipos, eq_loc, g_loc, g_vis)
        actualizar_estadisticas_equipo(tabla_equipos, eq_vis, g_vis, g_loc)

    # Convertir a DataFrame de pandas
    df = pd.DataFrame.from_dict(tabla_equipos, orient="index")
    df.index.name = "equipo"

    # Estructura exacta de columnas pedida en el ejercicio
    columnas_ordenadas = ["goles_a_favor", "goles_en_contra", "ganados", "empatados", "perdidos", "puntos"]
    df = df[columnas_ordenadas]

    # Ordenar de mayor a menor por puntos
    df_ordenado = df.sort_values(by="puntos", ascending=False)
    return df_ordenado


def mostrar_informe_deportivo(df_clasificacion):
    """Imprime el resultado exactamente como lo pide el enunciado."""
    # 1. Mostrar la tabla formateada
    print("DATAFRAME DE CLASIFICACIÓN FINAL")
    print(df_clasificacion.to_string())


    # 2. Identificar equipo más goleador y más goleado
    max_goles_favor = df_clasificacion["goles_a_favor"].max()
    mas_goleador = df_clasificacion[df_clasificacion["goles_a_favor"] == max_goles_favor].index[0]

    max_goles_contra = df_clasificacion["goles_en_contra"].max()
    mas_goleado = df_clasificacion[df_clasificacion["goles_en_contra"] == max_goles_contra].index[0]

    # 3. Formato del texto esperado según el resultado de la plataforma
    print(f"Equipo más goleador: {mas_goleador} ({max_goles_favor} goles)")
    print(f"Equipo más goleado:  {mas_goleado} ({max_goles_contra} goles)")
    print("\nClasificación final:")

    posicion = 1
    for equipo, fila in df_clasificacion.iterrows():
        print(f"{posicion}. {equipo} - {fila['puntos']} puntos")
        posicion += 1



def principal_liga():
    """Función principal orquestadora."""
    lineas = leer_archivo_partidos("historic partits.txt")
    if lineas:
        df_clasificacion = generar_clasificacion_dataframe(lineas)
        mostrar_informe_deportivo(df_clasificacion)
        return df_clasificacion

# Ejecutar análisis
df_resultado_liga = principal_liga()
```
